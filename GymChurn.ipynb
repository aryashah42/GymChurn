# 4: Random Forest

```python
import pandas as pd
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, roc_curve, auc
from sklearn.model_selection import train_test_split, StratifiedKFold
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('gym_churn_us.csv')
x = df.drop('Churn', axis=1)
y = df['Churn']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)
y_pred_test = rf_model.predict(x_test)

cm_test = confusion_matrix(y_test, y_pred_test)
accuracy_test = accuracy_score(y_test, y_pred_test)


print('Confusion Matrix: ')
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
print('Prediction Accuracy: ', accuracy_test)
print('Prediction Error: ', 1-accuracy_test)

TP = cm_test[1][1]
TN = cm_test[0][0]
FP = cm_test[0][1]
FN = cm_test[1][0]

print('True Positive Rate: ', TP / (TP + FN))
print('True Negative Rate: ', TN / (TN + FP))

f1_score = f1_score(y_test, y_pred_test)
print('F1 Score: ', f1_score)
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True)
auc_scores = []
accuracy_scores = []

fig, axes = plt.subplots(1, 5, figsize=(20, 5))
fig.suptitle('ROC Curves for Each Fold')

for fold, (train_index, test_index) in enumerate(cv.split(x_test, y_test)):
    x_train_fold, x_test_fold = x_test.iloc[train_index], x_test.iloc[test_index]
    y_train_fold, y_test_fold = y_test.iloc[train_index], y_test.iloc[test_index]

    rf_model.fit(x_train_fold, y_train_fold)
    y_probs = rf_model.predict_proba(x_test_fold)[:, 1]

    # Calculating ROC curve and AUC and accuracy for current fold
    fpr, tpr, thresholds = roc_curve(y_test_fold, y_probs)
    auc_score = auc(fpr, tpr)
    auc_scores.append(auc_score)
    y_pred = rf_model.predict(x_test_fold)
    accuracy = accuracy_score(y_test_fold, y_pred)
    accuracy_scores.append(accuracy)


    ax = axes[fold]
    ax.plot(fpr, tpr, label=f'Fold {fold + 1} (AUC = {auc_score:.2f})')
    ax.plot([0, 1], [0, 1], 'k--')
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.set_title(f'Fold {fold + 1}')
    ax.legend(loc='lower right')
    print(f'Fold {fold + 1} Accuracy: {accuracy:.2f}')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

# 6: Neural Network


```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt

from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# Loading data
df = pd.read_csv('gym_churn_us.csv')
X = df.drop('Churn', axis=1)
y = df['Churn']
categorical_cols = ['gender', 'Near_Location', 'Partner', 'Promo_friends', 'Phone']
numerical_cols = ['Age', 'Avg_additional_charges_total', 'Month_to_end_contract',
                  'Lifetime', 'Avg_class_frequency_total', 'Avg_class_frequency_current_month']

# Preprocessing data
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), numerical_cols),
    ('cat', OneHotEncoder(), categorical_cols)
])

X_processed = preprocessor.fit_transform(X)

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42)

# Model Architecture
model = Sequential([
    Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')  # Binary classification output
])

# Training
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history = model.fit(X_train, y_train, validation_split=0.2, epochs=20, batch_size=32)

# Evaluating
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {test_accuracy:.2f}")

plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
# Tuning Neural Network

# Calculate class weights
class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
y_train_np = y_train.to_numpy()
class_weights_dict = dict(enumerate(class_weights))

# Model Architecture
model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    BatchNormalization(),
    Dropout(0.2),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Model Training
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

history = model.fit(X_train, y_train_np, validation_split=0.2, epochs=100, batch_size=32,
                    callbacks=early_stopping, class_weight=class_weights_dict)

# Evaluating the model
y_test_np = y_test.to_numpy().reshape(-1) # Ensure y_test is a 1D numpy array
test_loss, test_accuracy = model.evaluate(X_test, y_test_np)
print(f"Test Accuracy: {test_accuracy:.2f}")

# Plotting
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

### 2) Performance  
To assess the performance of our Neural Network, we are using accuracy and binary-cross entropy. Accuracy is the proportion of correctly classified examples out of the total number of examples. We used it because it is a simple and intuitive metric to understand overall performance, especially in balanced datasets where the classes (e.g., churn vs. not churn) are roughly equal in size. Additionally, we also used binary-cross entropy, which is a loss function that measures the difference between the predicted probabilities and the actual binary outcomes. We used binary-cross entropy because it is well-suited for binary classification tasks. It penalizes incorrect predictions more heavily, especially when the predicted probability is far from the actual class.

### 3) Hyperparameters
For data preprocessing, we normalized the numerical features, one-hot encoded for categorical variables, and then we split our data into training, test, and validation test sets (80-20). For our model architecture, we used hidden layers containing 64, and 32 hidden nodes. We then had an output layer with 1 neuron and sigmoid activation for binary classification. For training, we set the number of epochs to 20, batch size to 32 (the largest we could) to balance training speed and generalization, and 20% of the training data for validation.

For our hyperparameters, we used the Adam optimizer to help adapt the learning rate dynamically. For more robust tuning, we could have also used tools like learning rate schedulers or grid search.
